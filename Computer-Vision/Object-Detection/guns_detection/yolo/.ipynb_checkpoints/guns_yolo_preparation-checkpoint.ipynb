```python
def create_dataset():
    data_folder = 'guns'
    images = os.listdir(os.path.join(data_folder, 'Images'))
    train_images, test_images = train_test_split(images, test_size=0.2, shuffle=True, random_state=42)
    for images, folder in [(train_images, train_folder), (test_images, test_folder)]:
        for image in tqdm(images):
            to_save = Image.open(os.path.join(data_folder, 'Images', image))
            h, w, c = np.array(to_save).shape
            to_save.save(os.path.join(folder, 'images', image))
            
            num = image.split('.')[0]
            with open(os.path.join(data_folder, 'Labels', f'{num}.txt'), 'r') as f:
                lines = f.readlines()
            bboxes = lines[1:]
            with open(os.path.join(folder, 'labels', f'{num}.txt'), 'w') as f:
                for bbox in bboxes:
                    x1, y1, x2, y2 = map(int, bbox.split())
                    xc = (x1 + x2) / (2 * w)
                    yc = (y1 + y2) / (2 * h)
                    bb_w = (x2 - x1) / w
                    bb_h = (y2 - y1) / h
                    f.write(f'0 {xc} {yc} {bb_w} {bb_h}\n')

create_dataset()
```

```python
with open('data.yaml','w') as f:
    data = """train: dataset/train\nval: dataset/test\nnc: 1\nnames: [\n'gun'\n]"""
    f.write(data)
    f.close()
```

```python
import gc
gc.collect()
torch.cuda.empty_cache()
```

```python
model = YOLO("yolov8n.pt") 

results = model.train(
    data="data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    name="my_custom_training"
)
```
